What is EDA?
=
EDA stands for Exploratory Data Analysis.

EDA is the process of examining, cleaning, summarizing, and visualizing a dataset to understand its patterns, relationships, and problems before applying Machine Learning.

Why do we perform EDA?
- To understand the dataset.
- To identify missing values and duplicate records.
- To detect outliers.
- To understand data distributions.
- To find relationships between variables.
- To identify useful features for Machine Learning.

Why is EDA performed before Machine Learning?
=

EDA is performed before Machine Learning to understand the dataset, identify missing values, duplicates, and outliers, analyze feature relationships, and discover patterns. It helps us clean and prepare the data, select relevant features, and build a more reliable Machine Learning model.

In simple words: EDA helps us understand our data before teaching a Machine Learning model to learn from it.


Difference between univariate, bivariate, and multivariate analysis.
=

Univariate analysis studies one variable to understand its distribution. Bivariate analysis studies two variables to identify their relationship. Multivariate analysis studies three or more variables to understand complex relationships and patterns in a dataset. These techniques are important in EDA before building Machine Learning models.

How do you identify missing values?
=

We use Pandas functions such as isnull() or isna() to identify missing values and sum() to count them in each column. This helps us decide how to handle missing data before training a Machine Learning model.

How do you identify duplicate records?
=

We use Pandas' duplicated() function to identify duplicate rows and duplicated().sum() to count them. We can use drop_duplicates() to remove duplicates when appropriate after checking the data.

What is an outlier?
=
An outlier is a data point that differs significantly from the other observations in a dataset. We can identify outliers using methods such as IQR, Z-score, box plots, and scatter plots. Detecting outliers helps us understand data quality and prepare data for Machine Learning.



Explain the IQR method.
=
The IQR method identifies potential outliers using the first quartile (Q1) and third quartile (Q3). We calculate IQR as Q3 − Q1. Values below Q1 − 1.5 × IQR or above Q3 + 1.5 × IQR are flagged as potential outliers. We investigate these values before deciding whether to keep or remove them.

What is skewness?
=
Skewness measures the asymmetry of a dataset's distribution. Positive skewness indicates a longer right tail, negative skewness indicates a longer left tail, and skewness near zero indicates an approximately symmetric distribution. In EDA, it helps us understand data distribution and decide whether transformations may be useful for Machine Learning.


What is correlation?
=


Correlation is a statistical measure that describes the strength and direction of the relationship between two variables. Its coefficient ranges from −1 to +1. Positive correlation means the variables tend to increase together, negative correlation means they tend to move in opposite directions, and zero correlation indicates no linear relationship.

Difference between correlation and causation.
=

Correlation indicates an association between two variables, whereas causation means a change in one variable directly produces a change in another. Correlation alone does not prove causation because the relationship may be influenced by other factors. In EDA, we identify correlations, but establishing causation requires additional evidence, such as controlled experiments or appropriate causal analysis.

What is multicollinearity?
=

Multicollinearity occurs when two or more independent variables in a dataset are highly correlated. It can make regression coefficients unstable and difficult to interpret. We can detect it using a correlation matrix or VIF and address it through feature selection or regularization.

How do you identify class imbalance?
=

I identify class imbalance using value_counts() and class percentages in Pandas, and visualize the distribution using a count plot. If one class has significantly more samples than the other, the dataset may be imbalanced. I then evaluate the model using precision, recall, F1-score, and a confusion matrix rather than relying only on accuracy.

Why should outliers not always be removed?
=

Outliers should not always be removed because they may represent genuine observations or important events, such as fraud or unusually large transactions. Removing them without investigation can cause information loss and negatively affect model performance. We should first identify the reason for each outlier and then decide whether to keep, correct, transform, or remove it.
