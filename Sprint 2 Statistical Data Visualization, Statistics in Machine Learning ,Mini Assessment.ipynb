Histogram
=    

Purpose
=


A histogram shows the distribution of numerical data by grouping values into intervals called bins.

When to use
=
To understand data distribution

To identify skewness

To see concentration of values

To detect possible outliers

Advantages
=
Easy to understand

Shows distribution clearly

Helps identify skewness

Limitations
=
Depends on bin size

Not suitable for categorical data

Different bin sizes can give different appearances

Interpretation
=
The histogram shows how frequently different mark ranges occur. If most values are concentrated around 60–70, that range represents 
the common performance.

```python
import matplotlib.pyplot as plt

marks = [45, 50, 52, 55, 60, 62, 65, 70, 72, 75, 80, 85]

plt.hist(marks, bins=5)
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.title("Distribution of Marks")
plt.show()
```

Explain the difference between Mean and Median
=

Mean: Sum of all values divided by the number of values.

Example: 10, 20, 30

Mean = (10 + 20 + 30) / 3 = 20

Median: The middle value after sorting the data.

Example: 10, 20, 30

Median = 20

Difference: Mean is affected by extreme values, while Median is less affected by outliers.

What is Variance?
=    

Variance measures how far the data values are spread from the mean.

Low variance → values are close to the mean.
    
High variance → values are widely spread.

Example:

2, 4, 6

Mean = 4

Variance measures the squared differences from 4.

What is Standard Deviation?
=
Standard deviation measures the amount of variation or spread in a dataset.

Small standard deviation → data is close to the mean.
    
Large standard deviation → data is more spread out.

Standard deviation is the square root of variance.

Explain Bayes' Theorem with an example
=    

Bayes' Theorem is used to calculate the probability of an event based on new evidence.

Formula:

P(A|B) = [P(B|A) × P(A)] / P(B)

Example:

Suppose:

Probability of having a disease = 1%
    
Probability of a positive test if a person has the disease = 90%
                                                           
Probability of a positive test = 5%

Then Bayes' theorem helps calculate:

Probability that a person has the disease given that the test is positive.




Difference between Population and Sample
=                                                           

Population: The complete group being studied.

Example: All students in a college.

Sample: A smaller group selected from the population.

Example: 100 students selected from the college.

Difference:

Population → entire group
    
Sample → subset of the population
    
Explain Correlation and Covariance
=    

Covariance: Shows the direction in which two variables change together.

Positive covariance → both tend to increase together.
Negative covariance → one tends to increase when the other decreases.

Correlation: Measures both the direction and strength of the relationship between two variables.

Correlation ranges from -1 to +1.



What is Skewness?
=    

Skewness measures the asymmetry of a distribution.

Types:

Positive skew: Tail is longer on the right.
    
Negative skew: Tail is longer on the left.
    
Zero skew: Distribution is approximately symmetric.
    
Explain IQR
=    

IQR (Interquartile Range) measures the spread of the middle 50% of data.

Formula:

IQR = Q3 - Q1

Where:

Q1 = First Quartile
Q3 = Third Quartile
Outlier rule

Lower limit:

Q1 - 1.5 × IQR

Upper limit:

Q3 + 1.5 × IQR

Values outside these limits are commonly considered outliers.

What is a p-value?
=    

A p-value helps determine whether the observed result provides evidence against the null hypothesis.

Small p-value → stronger evidence against the null hypothesis.
Large p-value → weaker evidence against the null hypothesis.

A commonly used significance level is 0.05.

Important: A p-value is not the probability that the null hypothesis is true.

Difference between Type I and Type II Error
=
    
Error	Meaning
    
Type I Error	Rejecting a true null hypothesis
    
Type II Error	Failing to reject a false null hypothesis

Simple example:

Type I: False positive
Type II: False negative
