What is a dataset?
=
A dataset is a collection of related data arranged in rows and columns.
    
Rows: Individual records, such as students.
    
Columns: Attributes describing each record, such as age, attendance, and study hours.

```python
 print("Dataset:")
print(df)
```

What is a feature?
=
A feature is an input variable used to understand or predict something.
    
Examples from this dataset:
=    
- Age
- Gender
- Study_Hours
- Attendance
- Course

```python
features = ["Age", "Gender", "Study_Hours", "Attendance", "Course"]

print("Features:", features)
```

What is a target?
=
A target is the output variable that a machine learning model tries to predict.
    
In this example, Passed is the target because we want to predict whether a student passes an exam.

Independent and dependent variables
=
    
- Independent variables (X): Inputs used for prediction.
    
- Dependent variable (y): The outcome we want to predict.

```python
X = df[["Age", "Gender", "Study_Hours", "Attendance", "Course"]]
y = df["Passed"]

print("Independent variables:")
print(X.head())

print("\nDependent variable:")
print(y.head())
```

Identify column categories
=

```python
numerical_columns = df.select_dtypes(include="number").columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category"]
).columns.tolist()

datetime_columns = df.select_dtypes(
    include=["datetime"]
).columns.tolist()

identifier_columns = [
    col for col in df.columns
    if col.lower().endswith("_id")
]

print("Numerical columns:", numerical_columns)
print("Categorical columns:", categorical_columns)
print("Date/time columns:", datetime_columns)
print("Identifier columns:", identifier_columns)
```

Identify potential target columns
=
A potential target is a column that represents the outcome you want to predict.
It cannot be identified reliably from data types alone; the problem statement determines the target.

Interpretation: 
=
Passed is a reasonable target for predicting whether a student passes or fails.
Student_ID should generally be excluded from model inputs because it identifies a record rather than measuring student performance.    

```python
potential_targets = ["Passed"]

for col in potential_targets:
    print(f"\nTarget column: {col}")
    print("Unique values:", df[col].unique())
    print("Value counts:")
    print(df[col].value_counts())
```

Create an inspection summary
=

This produces one summary table showing each column's data type, missing-value count, and number of unique values.    

```python
inspection_summary = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Non_Null_Count": df.notna().sum().values,
    "Missing_Count": df.isna().sum().values,
    "Unique_Values": df.nunique().values
})

inspection_summary
```
