Operators

Operators are symbols used to perform operations.
    
Arithmetic Operators :-  
    Arithmetic operators are used to perform basic mathematical operations in Python.
Python provides operators such as addition, subtraction, multiplication, division, floor division, modulus, and exponentiation.

```python
a = 10
b = 3

print(a + b)   # Addition: adds a and b
print(a - b)   # Subtraction: subtracts b from a
print(a * b)   # Multiplication: multiplies a and b
print(a / b)   # Division: divides a by b
print(a // b)  # Floor division: gives the whole-number quotient
print(a % b)   # Modulus: gives the remainder
print(a ** b)  # Exponentiation: a raised to the power of b
```

Comparison Operators:-

Comparison operators are used to compare two values. They return either True or False.
Python provides six main comparison operators: equal to, not equal to, greater than, less than, greater than or equal to, and less than or equal to. 
These operators are commonly used with conditional statements and loops.


```python
age = 20

print(age >= 18)
```

Logical Operators:-

Logical operators in Python are used to combine or reverse Boolean conditions. Python provides three logical operators: and, or, and not. 
They are mainly used in decision-making statements such as if, elif, and while

```python
# And
age = 25
salary = 30000

print(age >= 18 and salary >= 25000)
```

```python
# OR
print(age >= 18 or salary >= 50000)
```

```python
#Not
is_student = True

print(not is_student)
```

Assignment Operators:-

Assignment operators are used to store or update values in variables. 
The = operator assigns a value, while operators such as +=, -=, and *= perform an operation and then assign the result back to the variable


```python
Example:-

x = 10

x += 5
print(x)   # 15

x -= 3
print(x)   # 12

x *= 2
print(x)   # 24

x /= 4
print(x)   # 6.0
```

Input and Output

Input and Output (I/O) are basic concepts in Python. 
They allow a program to receive data from the user and display results to the user

Input

Input means taking data from the user while the program is running.
Python provides the input() function for this


```python
# Example:-

name = input("Enter your name: ")

print("Hello", name)
```

Output

Output means displaying information or results to the user.

Python uses the print() function for output.


```python
#Example:-
name = "Ram"
age = 25

print(name)
print(age)
```

Taking Multiple Inputs:-

we can take multiple values from the user.



```python
name = input("Enter your name: ")
age = int(input("Enter your age: "))

print("Name:", name)
print("Age:", age)
```

Enter your name: Ram
Enter your age: 25
Name: Ram
Age: 25

f-strings

A cleaner way to display variables is using an f-string
The f tells Python that variables can be placed inside {}.

```python
name = "Ram"
age = 25

print(f"My name is {name} and I am {age} years old.")
```

My name is Ram and I am 25 years old.
