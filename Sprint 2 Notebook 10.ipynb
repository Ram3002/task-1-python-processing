Notebook 10 – Hypothesis Testing
=
1. Null Hypothesis (H₀)
=

Definition:
=
Null Hypothesis means there is no significant difference or effect.

Example:
=
A company introduces a new training program.

H₀: New training does not improve employee performance.
    
2. Alternative Hypothesis (H₁)
=   

Definition:
=
Alternative Hypothesis means there is a significant difference or effect.

Example:
=

H₁: New training improves employee performance.
    
3. p-value
=    

Definition:
=
p-value tells us how strong the evidence is against H₀.

Simple rule:
=

p-value < 0.05 → Reject H₀

p-value ≥ 0.05 → Fail to reject H₀

Interpretation:
=
p = 0.03 means the observed result would be relatively unlikely if H₀ were true.

```python
p_value = 0.03

if p_value < 0.05:
    print("Reject Null Hypothesis")
else:
    print("Fail to Reject Null Hypothesis")

    
```

Confidence Interval
=

Definition:
=
A confidence interval gives a range of plausible values for a population parameter.

Example:
=    

Average customer spending = ₹500
95% CI = ₹450 to ₹550

This means the estimated population mean is supported by the interval ₹450–₹550 at the 95% confidence level.

5. Significance Level (α)
=

Definition:
=
Significance level is the threshold used to decide whether a result is statistically significant.

Common value:
=    

α = 0.05

If:

p-value < α

→ Reject H₀.

6. Type I Error
=

Definition:
=
Rejecting H₀ when H₀ is actually true.

Simple example:
=    
A medical test says a healthy person has a disease.

Also called: False Positive.

7. Type II Error
=

Definition:
=
Failing to reject H₀ when H₀ is actually false.

Simple example:
=
A medical test says a person with a disease is healthy.

8. One-Tailed Test
=

Used when we are interested in one direction.

Example:
=    

Does the new machine increase production?

H₀: μ ≤ 100
H₁: μ > 100

This is a right-tailed test.

9. Two-Tailed Test
=

Used when we want to check whether there is a difference in either direction.

Example:
=    

H₀: μ = 100
H₁: μ ≠ 100

It checks whether the mean is significantly higher or lower than 100.    

Notebook 11 – Statistical Tests
=
1. T-Test

Definition
=

A t-test compares means, especially when the population standard deviation is unknown.

Assumptions
=
Data is approximately normally distributed.
    
Observations are independent.
    
For independent two-sample t-tests, the equal-variance assumption depends on the version used.
    
When to use
=

When comparing the means of one or two groups, particularly with smaller samples.

When not to use
=

Don't use it blindly for strongly non-normal data, severe outliers, or dependent observations.

Business Example
=

Compare average sales before and after a marketing campaign.

Python Example

```python
from scipy.stats import ttest_ind

group1 = [20, 22, 21, 19, 23]
group2 = [25, 27, 26, 24, 28]

result = ttest_ind(group1, group2)

print(result)
```

Interpretation
=
p-value < 0.05 → Significant difference
p-value >= 0.05 → No statistically significant evidence of a difference

Z-Test
=
Definition
=
A z-test is used to test a population mean or proportion when the assumptions for a normal approximation are appropriate and the population standard deviation is known for the classical one-sample mean z-test.

Assumptions
=    
Independent observations.
    
Population standard deviation known for the classical one-sample mean test.
    
Sampling distribution is approximately normal.
    
When to use
=
Large samples or situations where population standard deviation is known.

When not to use
=    

Don't use the classical z-test when the required assumptions are not satisfied.

Business Example
=    

Check whether average delivery time differs from a known target of 30 minutes.

Python Example

```python
from statsmodels.stats.weightstats import ztest

data = [28, 31, 29, 30, 32, 27, 29]

stat, p = ztest(data, value=30)

print("Z-statistic:", stat)
print("p-value:", p)
```

Interpretation
=
p < 0.05 → Evidence against H₀
p >= 0.05 → Insufficient evidence against H₀

Chi-Square Test
=

Definition
=    

Chi-Square test is mainly used for categorical data to test relationships between categorical variables or compare observed and expected counts.

Assumptions
=    
Data consists of counts/frequencies.
    
Observations are independent.
    
Expected frequencies should generally be sufficiently large.
    
When to use
=

Example:
=

Is customer preference related to age group?

When not to use
=    

Don't use it for continuous measurements directly or when expected counts are too small for the approximation.

Business Example
=

Check whether product preference depends on customer gender

Interpretation
=
p < 0.05 → Evidence of an association
p >= 0.05 → Insufficient evidence of an association

```python
from scipy.stats import chi2_contingency

data = [
    [40, 20],
    [30, 30]
]

chi2, p, dof, expected = chi2_contingency(data)

print("Chi-Square:", chi2)
print("p-value:", p)
```

ANOVA
=
Definition
=

ANOVA compares the means of three or more groups.


Assumptions
=
Independent observations.
    
Approximately normal residuals within groups.
    
Similar variances across groups for standard one-way ANOVA
    
When to use
=                                

Compare average sales from three different regions.

When not to use
=    

Don't use standard ANOVA when its assumptions are seriously violated without considering alternatives.

Business Example
=

Compare sales from:
=    

Chennai
    
Bangalore
    
Hyderabad
    
Python Example    

```python
from scipy.stats import f_oneway

chennai = [100, 110, 105, 108]
bangalore = [120, 125, 122, 128]
hyderabad = [90, 95, 92, 94]

result = f_oneway(chennai, bangalore, hyderabad)

print("F-statistic:", result.statistic)
print("p-value:", result.pvalue)
```

Mann-Whitney U Test
=
Definition
=    

A non-parametric test used to compare two independent groups, especially when the normality assumption for a t-test is questionable.

Assumptions
=    
Two independent groups.
    
Outcome is at least ordinal.
    
Observations are independent.
    
When to use
=

When data is ordinal or heavily non-normal and you want to compare two independent groups.

When not to use
=    

Don't use it for paired/dependent observations.

Business Example

Compare customer satisfaction scores between two stores.

Interpretation
=
p < 0.05 → Evidence of a difference between groups
p >= 0.05 → Insufficient evidence of a difference    

Python Example

```python
from scipy.stats import mannwhitneyu

store_a = [3, 4, 2, 5, 3]
store_b = [5, 5, 4, 4, 5]

stat, p = mannwhitneyu(store_a, store_b)

print("U-statistic:", stat)
print("p-value:", p)
```

Shapiro-Wilk Test
=
Definition
=    

Shapiro-Wilk test checks whether data is consistent with a normal distribution.

Assumptions
=
Observations should be independent.
    
Appropriate for continuous data.
    
When to use
=

To assess normality before applying methods that assume normally distributed data.

When not to use
=    

Don't treat the p-value as the only evidence of normality, especially with very large datasets. Also consider a histogram or Q-Q plot.

Business Example
=

Check whether employee salaries in a sample are approximately normally distributed.

Interpretation
=
p < 0.05 → Evidence against normality
p >= 0.05 → Insufficient evidence against normality

```python
from scipy.stats import shapiro

data = [45, 48, 50, 52, 47, 49, 51]

stat, p = shapiro(data)

print("Statistic:", stat)
print("p-value:", p)
```

Multicollinearity
=
Definition
=
Multicollinearity occurs when two or more independent features are highly related.

Example:
=
Age ↔ Years of Experience

Both may contain similar information.

Problem

It can make regression coefficients unstable and harder to interpret.

3. Variance Inflation Factor (VIF)
=
Definition
=
VIF measures how much the variance of an estimated regression coefficient is inflated because of relationships among predictors.

Formula:

       VIF = 1 / (1 - R²)     

```python
from statsmodels.stats.outliers_influence import variance_inflation_factor

X = data[["Age", "Salary", "Experience"]]

for i in range(X.shape[1]):
    print(
        X.columns[i],
        variance_inflation_factor(X.values, i)
    )
```

Simple Interpretation
=

Higher VIF indicates stronger multicollinearity concerns.

A common practical rule is:
=

VIF > 5  → investigate

VIF > 10 → strong concern in many applications

These are rules of thumb, not universal laws.
