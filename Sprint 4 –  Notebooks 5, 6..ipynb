Notebook 6 – Multivariate Analysis
=

What is Multivariate Analysis?
=
    
Multivariate analysis studies relationships among three or more variables at the same time.
    
Example:
=
Analyze how age, income, spending, and salary relate to one another.


Correlation matrix
=    
A correlation matrix measures the linear relationship between numerical variables. Values range from -1 to +1.
    
Correlation	Meaning
=
+1	Perfect positive linear relationship

Near +1	Strong positive relationship

Near 0	Weak or no linear relationship
                           
Near -1	Strong negative relationship

-1	Perfect negative linear relationship    

```python
numerical_df = df[
    ["Age", "Salary", "Income", "Spending", "Sales"]
]

correlation_matrix = numerical_df.corr()

print(correlation_matrix.round(2))
```

Heatmap
=

A heatmap displays correlation values using colors, making strong and weak relationships easier to identify.

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    vmin=-1,
    vmax=1
)

plt.title("Correlation Matrix Heatmap")
plt.show()
```

Observation:
=

Look for values close to +1 or -1. Values near zero indicate weak linear relationships. Correlation does not establish causation

pair plot
=
A pair plot compares multiple numerical variables using scatter plots and distribution plots.

```python
sns.pairplot(
    numerical_df,
    diag_kind="hist"
)

plt.show()
```

Observation:
=

     Look for upward trends, downward trends, clusters, unusual points, and nonlinear patterns.
    
Multiple feature relationships
=
Suppose we want to examine how age, salary, and spending relate to one another.

```python
sns.scatterplot(
    data=df,
    x="Age",
    y="Spending",
    hue="Department",
    size="Salary"
)

plt.title("Age vs Spending by Department")
plt.show()
```

Here:
=
- X-axis represents age.
    
- Y-axis represents spending.
    
- Color represents department.
    
- Point size represents salary.
    
This helps investigate several features together.
    
Identify multicollinearity
=
    
Multicollinearity occurs when two or more independent features are strongly correlated.

```python
corr = numerical_df.corr()

high_corr = corr.where(
    corr.abs() > 0.80
)

print(high_corr)
```

This displays correlations above an absolute value of 0.80, including diagonal values and duplicated pairs.
    
To list only distinct feature pairs:

```python
import numpy as np

upper_triangle = corr.where(
    np.triu(np.ones(corr.shape), k=1).astype(bool)
)

print(upper_triangle.stack().loc[
    lambda x: x.abs() > 0.80
])
```
