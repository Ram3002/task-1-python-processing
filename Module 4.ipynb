 Log Levels
 =

Python logging provides different log levels.

They tell us how important a message is.

The main levels are:

DEBUG
INFO
WARNING
ERROR
CRITICAL

4.1 DEBUG
=


Use DEBUG when you need detailed information about what the program is doing internally

Example

```python
username = "Ram"

  logging.debug(f"Processing user: {username}")
```

4.2 INFO
=

INFO indicates that something normal and expected happened.

```python
logging.info("User logged in")
logging.info("Database connected")
logging.info("File processed successfully")
```

4.3 WARNING
=

Indicates a potential problem, but the program can still continue.

A warning does not necessarily mean the program has failed.    

```python
age = 17

if age < 18:
    logging.warning("User is under 18")
```

4.4 ERROR
=
ERROR indicates that something went wrong and an operation could not be completed successfully

```python
try:
    with open("data.txt") as file:
        data = file.read()
        
    except FileNotFoundError:
        logging.error("data.txt was not found")
```

4.5 CRITICAL
=

CRITICAL represents a very serious problem that may prevent the application from continuing.

Log Formatters
======

A formatter determines how a log message should look.

Without a formatter, the message may be very simple.

With a formatter, we can include:

Date and time,
Log level,
Logger name,
Message,
File name,
Line number,


```python
import logging

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)
```

File Handlers
=

A handler determines where log messages should go.

A FileHandler sends logs to a file.

Example

```python
import logging

file_handler = logging.FileHandler("application.log")

logging.basicConfig(
    level=logging.INFO,
    handlers=[file_handler]
)

logging.info("Application started")
```

Why use FileHandler?
=

It allows us to keep a permanent record of application activity.

Useful for:

Server applications/
Database applications/
Background processes/
Error investigation/
Production systems.

Console Handlers
=

A ConsoleHandler sends log messages to the console/terminal.

In Python, this is provided by StreamHandler.

```python
import logging

console_handler = logging.StreamHandler()
```

File Handler vs Console Handler
=

We can use both at the same time.

                 ┌──------------ Console
Application ─── Logger-----------Log File
                              

```python
import logging

logger = logging.getLogger("myapp")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()
file_handler = logging.FileHandler("application.log")

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.info("Application started")
```

Now the log can appear on the console and be stored in the file.

Rotating Logs
=


One problem with file logging is that the log file can become very large.

Imagine an application running for months:

application.log

This is not desirable.

Log rotation solves this problem by creating new log files when the current file reaches a certain size or after a certain amount of time.


RotatingFileHandler
=

means the file can reach approximately 1 MB before rotation.

When the current log becomes too large, older logs are rotated
    

Time-Based Log Rotation
=

We can also rotate logs based on time.
This creates a new log file periodically, here around midnight, while retaining a specified number of backups.

This is useful for applications where you want one log file per day.

```python
from logging.handlers import TimedRotatingFileHandler

handler = TimedRotatingFileHandler(
    "application.log",
    when="midnight",
    backupCount=7
)   
```

Complete Logging Example
=

Here is a good example combining several concepts:

This demonstrates:

Logger/
Log levels/
Console handler/
File handler/
Formatter/

```python
import logging

logger = logging.getLogger("myapp")
logger.setLevel(logging.DEBUG)

console_handler = logging.StreamHandler()
file_handler = logging.FileHandler("application.log")

formatter = logging.Formatter(
    "%(asctime)s - %(levelname)s - %(message)s"
)

console_handler.setFormatter(formatter)
file_handler.setFormatter(formatter)

logger.addHandler(console_handler)
logger.addHandler(file_handler)

logger.debug("Debug message")
logger.info("Application started")
logger.warning("Low disk space")
logger.error("File could not be opened")
logger.critical("Critical system failure")
```

Understanding Logger, Handler and Formatter
=
These three concepts are very important.

Think of logging like a post office.

Logger
=
The Logger creates the log message.

Handler
=
The Handler decides where the message goes.


Formatter
=
The Formatter decides how the message looks.


Overall flow
=    
Application
     ↓
   Logger
     ↓
   Handler
   ↙     ↘
Console   File
     ↓
 Formatter

A more accurate model is that the logger creates a LogRecord, handlers receive it, and each handler can apply its own formatter.    
    

Logging Best Practices
=

Following good logging practices makes applications easier to maintain.

1. Use appropriate log levels

Don't use ERROR for everything.

Bad:

  logging.error("Application started")

Better:

   logging.info("Application started") 

Don't log sensitive information
=
Never log things such as:

Passwords/
Credit card numbers/
API keys/
Authentication tokens/
Private personal information.

Bad:
=    

logging.info(f"User password: {password}")

This can create a security risk.

Use meaningful messages
=
    
Bad:
=
logging.error("Error")

Better:
=
logging.error("Failed to connect to the database")

The second message tells us what actually happened.    

Include useful context
=
Instead of:

logging.info("Processing")

Use:
=
logging.info("Processing customer ID: %s", customer_id)

This makes troubleshooting easier.

Use logger.exception() inside exception handling
=
When handling an exception:

try:
    result = 10 / 0
except Exception:
    logger.exception("Calculation failed")

logger.exception() records the error along with the traceback, which is extremely useful for debugging.

Use log rotation
=
For applications that generate many logs, use:

RotatingFileHandler

or:

TimedRotatingFileHandler

This prevents log files from growing indefinitely.

CSV Files
===

CSV means Comma-Separated Values.

CSV files are commonly used for tabular data such as:

id,name,age

1,Ram,22

2,Arun,23

3,Priya,21

provides the built-in csv module

Writing a CSV File
=

```python



import csv

students = [
    [1, "Ram", 22],
    [2, "Arun", 23],
    [3, "Priya", 21]
]

with open("students.csv", "w", newline="") as file:
    writer = csv.writer(file)

    writer.writerow(["ID", "Name", "Age"])
    writer.writerows(students)
```

Reading CSV
=

```python
import csv

with open("students.csv", "r") as file:
    reader = csv.reader(file)

    for row in reader:
        print(row)
```

CSV Using Dictionary
=

A more readable approach is DictReader.

```python
import csv

with open("students.csv", "r") as file:
    reader = csv.DictReader(file)

    for row in reader:
        print(row["Name"], row["Age"])
```

JSON Files
=

JSON stands for JavaScript Object Notation.

It is commonly used for storing and exchanging structured data.

Example JSON:

{
    "name": "Ram",
    "age": 22,
    "course": "Python"
}

Writing JSON

```python
import json

student = {
    "name": "Ram",
    "age": 22,
    "course": "Python"
}

with open("student.json", "w") as file:
    json.dump(student, file, indent=4)
```

Reading JSON

```python
import json

with open("student.json", "r") as file:
    student = json.load(file)

print(student)
print(student["name"])
print(student["age"])
```

File Validation
=

File validation means checking whether a file is valid and safe to process before using it.

We can check:

Does the file exist?

Is it actually a file?

Is the extension correct?

Is the file empty?

Is the file readable?

Is the file size acceptable?

Simple Validation
=

```python
from pathlib import Path

file_path = Path("students.csv")

if not file_path.exists():
    print("File does not exist")

elif not file_path.is_file():
    print("This is not a file")

elif file_path.stat().st_size == 0:
    print("File is empty")

else:
    print("File is valid")
```

Validate File Extension
=

```python
from pathlib import Path

file_path = Path("students.csv")

if file_path.suffix == ".csv":
    print("CSV file")
else:
    print("Invalid file type")
```

Validate File Size
=

```python
from pathlib import Path

file_path = Path("students.csv")

max_size = 1024 * 1024  # 1 MB

if file_path.exists() and file_path.stat().st_size <= max_size:
    print("File size is acceptable")
else:
    print("File is too large")
```
