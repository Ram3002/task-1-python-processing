```python
class DataLoader:

    SUPPORTED_FORMATS = {".csv", ".json", ".txt"}

    def __init__(self, file_path):
        self.file_path = Path(file_path)

    def validate_path(self):

        if not self.file_path.exists(): 

            raise InvalidFileException(
                f"File does not exist: {self.file_path}"
            )

    def _load_csv(self):

        with self.file_path.open(
            mode="r",
            encoding="utf-8",
            newline=""
        ) as file:

            return list(csv.DictReader(file))

    def _load_json(self):

        with self.file_path.open(
            mode="r",
            encoding="utf-8"
        ) as file:

            return json.load(file)

    def _load_txt(self):

        return self.file_path.read_text(
            encoding="utf-8"
        )

    def load(self):

        self.validate_path()

        file_extension = self.file_path.suffix.lower()

        if file_extension not in self.SUPPORTED_FORMATS:

            raise UnsupportedFileFormatException(
                f"Unsupported file format: {file_extension}"
            )

        if file_extension == ".csv":
            return self._load_csv()

        elif file_extension == ".json":
            return self._load_json()

        elif file_extension == ".txt":
            return self._load_txt()

#Class DataLoader
# To provide one reusable place for loading different types of files.
# Support format CSV,Json & TXT 
#.csv  → supported
#.json → supported
#.txt  → supported
#.xlsx → not supported
#.pdf  → not supported
#.xml  → not supported
#SUPPORTED_FORMATS = {".csv", ".json", ".txt"}
#This is a class variable. 
# def validate_path- A method that checks whether the file actually exists on disk using Path.exists
  #(). If it doesn't, it raises a custom exception InvalidFileException with a descriptive f-string
  #message including the offending path
# def load CSV-A Define a method called _load_csv that belongs to this object and will load a CSV file.
# def load Json- Its purpose is to load JSON data
#def  load TXT-Its job is to load a text file.
# def load(self)- Read the entire text file as a string and return it.”

           


```

```python
#Data Writer
import csv
import json
from pathlib import Path


class InvalidFileException(Exception):
    pass


class UnsupportedFileFormatException(Exception):
    pass


class DataLoader:

    SUPPORTED_FORMATS = {".csv", ".json", ".txt"}

    def __init__(self, file_path):
        self.file_path = Path(file_path)

    def validate_path(self):

        if not self.file_path.exists():

            raise InvalidFileException(
                f"File does not exist: {self.file_path}"
            )

    def _load_csv(self):

        with self.file_path.open(
            mode="r",
            encoding="utf-8",
            newline=""
        ) as file:

            return list(csv.DictReader(file))

    def _load_json(self):

        with self.file_path.open(
            mode="r",
            encoding="utf-8"
        ) as file:

            return json.load(file)

    def _load_txt(self):

        return self.file_path.read_text(
            encoding="utf-8"
        )

    def load(self):

        self.validate_path()

        file_extension = self.file_path.suffix.lower()

        if file_extension not in self.SUPPORTED_FORMATS:

            raise UnsupportedFileFormatException(
                f"Unsupported file format: {file_extension}"
            )

        if file_extension == ".csv":
            return self._load_csv()

        elif file_extension == ".json":
            return self._load_json()

        elif file_extension == ".txt":
            return self._load_txt()
```
