What is Pandas?
=

Pandas is a Python library used for data manipulation, data analysis, and data cleaning.

```python
import pandas as pd
```

Why Pandas?
=

Pandas is useful for:
=

Handling structured data

Data analysis

Example:
=

```python
data = {
    "Name": ["Ram", "John", "Sam"],
    "Age": [25, 30, 28]
}

df = pd.DataFrame(data)
print(df)
```

Series
=

A Series is a one-dimensional labeled array.

```python
import pandas as pd

s = pd.Series([10, 20, 30, 40])

print(s)
```

Series with custom index
=

```python
s = pd.Series(
    [90, 85, 95],
    index=["Math", "Science", "English"]
)

print(s)
```

DataFrame
=

A DataFrame is a two-dimensional table containing rows and columns.

```python
data = {
    "Name": ["Ram", "John", "Sam"],
    "Age": [25, 30, 28],
    "Salary": [30000, 40000, 35000]
}

df = pd.DataFrame(data)

print(df)
```

Excel Files
=

=

df = pd.read_excel("data.xlsx")

JSON Files

JSON is commonly used for structured and API data.

Write

Parquet Files
=

Parquet is a columnar storage format commonly used in data engineering and big-data workflows.

Read
=

Write
=

df.to_parquet("output.parquet")

Pickle Files
=

Pickle stores Python objects in a serialized format.

Read
=
df = pd.read_pickle("data.pkl")

Write
=

df.to_pickle("data.pkl")

SQL Tables
=

Pandas can read data from SQL databases.

```python
import pandas as pd
import sqlite3

connection = sqlite3.connect("company.db")

df = pd.read_sql(
    "SELECT * FROM employees",
    connection
)

print(df)
```

Write DataFrame to SQL:
=

```python
df.to_sql(
    "employees",
    connection,
    if_exists="replace",
    index=False
)
```

Reading Multiple Files
=

For multiple CSV files:

This is useful when data is divided into multiple files.    

```python
from pathlib import Path
import pandas as pd

files = Path("data").glob("*.csv")

df = pd.concat(
    [pd.read_csv(file) for file in files],
    ignore_index=True
)

print(df)
```

File Path Handling
=

Use pathlib for platform-independent file paths.

```python
from pathlib import Path

path = Path("data") / "employees.csv"

df = pd.read_csv(path)
```

Notebook 6 – Data Inspection
=

Data inspection means understanding the structure, data types, size, and contents of a dataset before performing analysis or ML operations.

Let's use this DataFrame:

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Ram", "John", "Sam", "Ram", "David"],
    "Age": [25, 30, 28, 25, 35],
    "Department": ["IT", "HR", "IT", "IT", "Finance"],
    "Salary": [30000, 40000, 35000, 30000, 50000]
})
```
