Copy vs View
=
Copy
=    

A copy creates a completely independent array.

arr = np.array([1, 2, 3])
copy_arr = arr.copy()

copy_arr[0] = 100

print(arr)

Output:
=

[1 2 3]
    
View
=    

A view shares the original array's data.

arr = np.array([1, 2, 3])
view_arr = arr.view()

view_arr[0] = 100

print(arr)

Output:
=

[100   2   3]

Key difference: Copy has independent data; View shares data.

Standard Deviation
=

Standard deviation measures the spread of values around the mean.

arr = np.array([10, 20, 30])

\(\sigma = \sqrt{\frac{1}{N}\sum_{i=1}^{N}(x_i-\mu)^2}\)

The points have mean μ = 5 and σ = 1.5.


Key point: Lower standard deviation means values are closer to the mean.

Notebook 3 – NumPy Random Module

Random Numbers
=

Definition:
=
NumPy provides the numpy.random module to generate random numbers for simulations, testing, sampling, and machine learning.

```python
import numpy as np

numbers = np.random.rand(5)
print(numbers)
```

Random Seed
=

Definition:
=
A random seed makes random results reproducible. The same seed produces the same sequence of random numbers.

Use:
=

Testing

Debugging

Machine learning experiments

Reproducible results    

```python
import numpy as np

np.random.seed(42)

print(np.random.rand(5))
```

Random Integers
=
np.random.randint() generates random integer values.

Syntax:
=

     np.random.randint(low, high, size)

```python
import numpy as np

numbers = np.random.randint(1, 10, size=5)

print(numbers)
```

Random Choice
=

np.random.choice() randomly selects values from an array.

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

result = np.random.choice(data)

print(result)
```

Select multiple values:

result = np.random.choice(data, size=3)

print(result)

Possible output:

[20 50 10]

Example: Randomly selecting customers for a survey.

Shuffle
=

np.random.shuffle() randomly rearranges the elements of an array in-place.

The original array itself is modified.    

```python
import numpy as np

data = np.array([1, 2, 3, 4, 5])

np.random.shuffle(data)

print(data)
```

Permutation
=

np.random.permutation() returns a new shuffled array.

| Feature        | `shuffle()`              | `permutation()`   |
| -------------- | ------------------------ | ----------------- |
| Original array | Modified                 | Not modified      |
| Returns        | `None`                   | New array         |
| Use            | Rearrange existing array | Get shuffled copy |


```python
import numpy as np

data = np.array([1, 2, 3, 4, 5])

result = np.random.permutation(data)

print(result)
print(data)
```

Random Distributions
=

A probability distribution describes how random values are likely to occur.

NumPy provides many distributions.

Normal Distribution
=


import numpy as np

data = np.random.normal(
    loc=50,
    scale=10,
    size=5
)

print(data)


loc → mean
scale → standard deviation
size → number of values

Example: Student exam scores.

Uniform Distribution
=

All values within a specified range have equal probability

data = np.random.uniform(
    low=0,
    high=10,
    size=5
)

print(data)




Binomial Distribution
=

Used when there are a fixed number of independent trials with two possible outcomes.

 data = np.random.binomial(
    n=10,
    p=0.5,
    size=5
)

print(data)  

Example: 
=

Number of successful outcomes in 10 attempts.


Poisson Distribution
=

Used to model the number of events occurring within a fixed interval.

data = np.random.poisson(
    lam=3,
    size=5
)

print(data)

Example:
=

Number of customers arriving at a store per hour.   

Exponential Distribution
=

Used to model the time between events.

data = np.random.exponential(
    scale=2,
    size=5
)

print(data)

Example:
=

Waiting time between customer arrivals.

Random functions are commonly used for sampling data, creating test datasets, simulations, initializing experiments, and splitting data randomly.
