Notebook 7 – Selecting & Filtering Data
=

loc
=

Select rows and columns using labels.


```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["A", "B", "C", "D"],
    "Age": [21, 25, 23, 30],
    "Salary": [25000, 35000, 30000, 45000]
})

print(df.loc[1])
print(df.loc[1:2, ["Name", "Salary"]])
```

iloc
=

Select rows and columns using integer positions.

```python
print(df.iloc[0])
print(df.iloc[1:3, 0:2])
```

Boolean Indexing
=

```python
result = df[df["Age"] > 23]
print(result)
```

Filtering Rows
=

```python
result = df[df["Salary"] >= 30000]
print(result)
```

query()
=

```python
result = df.query("Age > 23 and Salary > 30000")
print(result)
```

isin()
=

```python
result = df[df["Name"].isin(["A", "C"])]
print(result)
```

between()
=

```python
result = df[df["Age"].between(22, 28)]
print(result)
```

Sorting Data
=

```python
print(df.sort_values("Salary"))
print(df.sort_values("Salary", ascending=False))
```

```python
df["Salary_With_Bonus"] = df["Monthly_Salary"].apply(
    lambda x: x * 1.10
)

print(df[["Name", "Salary_With_Bonus"]])

Output:
```

map()
=

map() is commonly used with a Series to transform individual values.

```python
department_code = {
    "IT": 101,
    "HR": 102,
    "Sales": 103
}

df["Dept_Code"] = df["Department"].map(department_code)
print(df[["Name", "Department", "Dept_Code"]])
```

replace()
=

```python
df["Department"] = df["Department"].replace({
    "IT": "Technology"
})

print(df[["Name", "Department"]])
```

Notebook 9 – Grouping & Aggregation
=

```python
df = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Employee": ["A", "B", "C", "D", "E", "F"],
    "Salary": [30000, 40000, 25000, 35000, 28000, 32000]
})

print(df)
```

groupby()
=

Group employees by department.

```python
result = df.groupby("Department")["Salary"].mean()
print(result)
```

agg()
=

Perform multiple aggregations.

```python
result = df.groupby("Department")["Salary"].agg(
    ["sum", "mean", "min", "max"]
)

print(result)
```

transform()
=

Returns results aligned with the original DataFrame.

```python
df["Dept_Avg"] = df.groupby("Department")["Salary"].transform("mean")

print(df)
```

filter()
=

Filter groups based on a condition.

```python
result = df.groupby("Department").filter(
    lambda x: x["Salary"].mean() > 30000
)

print(result)
```

Pivot Tables
=

```python
pivot = pd.pivot_table(
    df,
    values="Salary",
    index="Department",
    aggfunc="mean"
)

print(pivot)
```

Crosstab
=

```python
data = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales", "Sales"],
    "Gender": ["M", "F", "F", "M", "M", "F"]
})

result = pd.crosstab(data["Department"], data["Gender"])
print(result)
```

Notebook 10 – Combining Datasets
=

```python
employees = pd.DataFrame({
    "Emp_ID": [1, 2, 3],
    "Name": ["A", "B", "C"]
})

salary = pd.DataFrame({
    "Emp_ID": [1, 2, 4],
    "Salary": [30000, 40000, 50000]
})
```

concat()
=

Combine DataFrames vertically.

```python
df1 = pd.DataFrame({
    "Name": ["A", "B"],
    "Age": [21, 25]
})

df2 = pd.DataFrame({
    "Name": ["C", "D"],
    "Age": [23, 30]
})

result = pd.concat([df1, df2], ignore_index=True)
print(result)
```

merge()
=

Combine datasets using a common key.

```python
result = pd.merge(
    employees,
    salary,
    on="Emp_ID",
    how="inner"
)

print(result)
```

join()
=

join() is mainly used to combine DataFrames based on their index.

```python
df1 = pd.DataFrame({
    "Name": ["A", "B", "C"]
}, index=[1, 2, 3])

df2 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

result = df1.join(df2)
print(result)
```

combine_first()
=

Fills missing values in one DataFrame using values from another DataFrame.

```python
df1 = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Salary": [30000, None, 35000]
})

df2 = pd.DataFrame({
    "Name": ["A", "B", "C"],
    "Salary": [32000, 40000, 36000]
})

result = df1.combine_first(df2)

print(result)
```
