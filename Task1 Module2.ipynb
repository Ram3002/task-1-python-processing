 Control Flow
 ============

Control flow means controlling the order in which Python statements are executed.

Normally, Python executes code from top to bottom. Control-flow statements allow us to make decisions, repeat code, or skip/stop execution.

if 
   
    Checks the first condition.
    
elif
    
    checks another condition when the previous condition is false
    
else

    Runs when none of the conditions are true


These statements are used for decision making    

```python
#Example:-

marks = 75

if marks >= 90:
    print("A+")
elif marks >= 60:
    print("A")
else:
    print("B")

```

for Loop
========

A for loop is used to repeat a block of code for each item in a sequence.

A for loop is commonly used with:

Lists

Tuples
    
Strings
    
Dictionaries
    
range()
    
Other iterable objects

```python
fruits = ["Apple", "Banana", "Mango"]

for fruit in fruits:
    print(fruit)
```

while Loop
==========

A while loop repeats code as long as a condition is true.

The condition is checked before every iteration    

```python
count = 1

while count <= 5:
    print(count)
    count += 1
```

break
=====

break immediately stops a loop.

When i becomes 5, the loop stops.    

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

continue
=========

continue skips the current iteration and moves to the next iteration.

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

The value 2 is skipped

pass

pass does nothing. It is used as a placeholder when Python requires a statement but you don't want to execute anything yet.
It is useful when developing code that you plan to implement later.

```python
age = 20

if age >= 18:
    pass
```

Recursive Functions
===========

A recursive function calls itself.
A recursive function needs a base condition to stop recursion.

```python
def factorial(n):
    if n == 0:
        return 1
    return n * factorial(n - 1)

print(factorial(5))
```

Decorators
=======

A decorator is a function that adds or modifies behavior of another function without changing its source code

@decorator applies the decorator to greet()

```python
def decorator(func):
    def wrapper():
        print("Before function")
        func()
        print("After function")
    return wrapper

@decorator
def greet():
    print("Hello")

greet()
```

Closures
=

A closure occurs when an inner function remembers variables from its outer function even after the outer function has finished

Hello

The inner function remembers message.

```python
def outer(message):
    def inner():
        print(message)

    return inner

func = outer("Hello")
func()
```

global and nonlocal
global
=
Used when we want to modify a global variable inside a function.

```python
count = 0

def update():
    global count
    count += 1

update()

print(count)
```

nonlocal

Used inside nested functions to modify a variable belonging to an enclosing function.

```python
def outer():
    count = 0

    def inner():
        nonlocal count
        count += 1

    inner()
    print(count)

outer()
```

Properties – @property
=

Properties allow methods to be accessed like attributes while providing control over getting or setting values.

```python
class Student:
    def __init__(self, age):
        self._age = age

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, value):
        if value < 0:
            raise ValueError("Age cannot be negative")
        self._age = value

student = Student(20)

print(student.age)

student.age = 21
```

Inheritance
=

Inheritance allows one class to reuse and extend another class.

    Dog inherits from Animal

```python
class Animal:
    def eat(self):
        print("Eating")

class Dog(Animal):
    def bark(self):
        print("Barking")

dog = Dog()

dog.eat()
dog.bark()
```

Multiple Inheritance
=

A class inherits from more than one parent class.

```python
class Father:
    def skill1(self):
        print("Driving")

class Mother:
    def skill2(self):
        print("Cooking")

class Child(Father, Mother):
    pass

child = Child()

child.skill1()
child.skill2()
```

Polymorphism
=

Polymorphism means one interface/name can have different behavior.

    The same method name sound() behaves differently

```python
class Dog:
    def sound(self):
        print("Bark")

class Cat:
    def sound(self):
        print("Meow")

for animal in [Dog(), Cat()]:
    animal.sound()
```

Method Overriding
=

A child class provides its own version of a parent method

The child's method overrides the parent's method.

```python
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        print("Bark")

dog = Dog()
dog.sound()
```

Method Overloading – Python Approach
=

Traditional method overloading means having multiple methods with the same name but different parameter lists.

Python does not support traditional compile-time method overloading in the same way as languages such as Java.

Instead, we can use default arguments or *args

```python
class Calculator:

    def add(self, a, b=0, c=0):
        return a + b + c

calculator = Calculator()

print(calculator.add(10))
print(calculator.add(10, 20))
print(calculator.add(10, 20, 30))
```

Composition
=

Composition represents a strong has-a relationship.

One object contains another object as a part of itself

A Car has an Engine.

```python
class Engine:
    def start(self):
        print("Engine started")

class Car:
    def __init__(self):
        self.engine = Engine()

car = Car()
car.engine.start()
```

Aggregation
=

Aggregation is also a has-a relationship, but the contained object can exist independently.

    The teacher can exist independently of the department

```python
class Teacher:
    pass

class Department:
    def __init__(self, teacher):
        self.teacher = teacher

teacher = Teacher()
department = Department(teacher)
```

Association
=

Association means two objects are related but don't necessarily own each other.

```python
class Student:
    pass

class Teacher:
    def teach(self, student):
        print("Teaching student")

student = Student()
teacher = Teacher()

teacher.teach(student)
```

Magic / Dunder Methods
=

Magic methods are special methods with names beginning and ending with __.

Examples:

__init__
__str__

They allow Python objects to work with built-in operations.

super()

super() is used to access functionality from a parent class.

Here super().__init__(name) calls the parent constructor.

```python
class Animal:
    def __init__(self, name):
        self.name = name

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)
        self.breed = breed
```

__str__()

Defines a user-friendly string representation of an object.

```python
class Student:
    def __init__(self, name):
        self.name = name

    def __str__(self):
        return f"Student: {self.name}"

student = Student("Ram")

print(student)
```

__repr__()

Provides a more developer-oriented representation of an object.

class Student:
    def __init__(self, name):
        self.name = name

    def __repr__(self):
        return f"Student(name={self.name!r})"

student = Student("Ram")

print(repr(student))

__len__()

Defines what happens when len() is called on an object.

```python
class Team:
    def __init__(self, members):
        self.members = members

    def __len__(self):
        return len(self.members)

team = Team(["A", "B", "C"])

print(len(team))
```

Singleton Pattern — Bonus

The Singleton Pattern ensures that a class has only one shared instance.

A simple example:

class Singleton:
    _instance = None

    def __new__(cls):
        if cls._instance is None:
            cls._instance = super().__new__(cls)
        return cls._instance

a = Singleton()
b = Singleton()

print(a is b)

Module 4 – Logging in Python
=

Logging is an important Python concept used to record what is happening inside an application. It helps developers understand program execution, find errors, monitor applications, and troubleshoot problems.

Python provides a built-in logging module, so we don't need to install an external package.

What is Logging?
==================


Logging is the process of recording information about the execution of a program.

For example, when an application starts, connects to a database, processes a file, or encounters an error, we can record these events in a log.

example
    import logging

logging.basicConfig(level=logging.INFO)

logging.info("Application started")
logging.warning("Low disk space")
logging.error("Database connection failed")

In simple words

Logging means keeping a record of important events, messages, warnings, and errors that happen while a program is running



Why Use Logging?
=


Logging is useful for several reasons.
    1. Debugging
    
It helps us find out where and why a problem occurred.
    2. Error tracking

We can record errors instead of relying only on the user seeing an error message.
    3. Monitoring

Logs allow us to monitor what an application is doing.
    4. Troubleshooting

If something goes wrong after the application has been running for several hours, logs can tell us what happened before the problem.

5. Permanent records

Unlike print(), logs can be saved to a file.
    6. Production applications

Logging is especially important for applications running on servers because developers may not be able to watch the application directly.
