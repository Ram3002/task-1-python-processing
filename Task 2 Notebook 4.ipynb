Normal Distribution
=
Definition
=
A Normal Distribution is a continuous probability distribution that is symmetric around its mean.

It has the famous bell-shaped curve.

Characteristics:
=    

Mean = Median = Mode

Symmetric around the mean

Most observations are near the mean

Fewer observations occur far from the mean

```python
import numpy as np
import matplotlib.pyplot as plt

data = np.random.normal(loc=50, scale=10, size=1000)

plt.hist(data, bins=30, density=True)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

Kurtosis
=
Kurtosis describes the shape of the tails of a distribution.

It helps identify whether a distribution has relatively:

Heavy tails

Normal-like tails

Light tails

| Type        | Meaning     |
| ----------- | ----------- |
| Mesokurtic  | Normal-like |
| Leptokurtic | Heavy tails |
| Platykurtic | Light tails |


```python
from scipy.stats import kurtosis

data = np.random.normal(0, 1, 1000)

print("Kurtosis:", kurtosis(data))
```

Business Example
=

Kurtosis can help identify whether extreme events occur more frequently than expected.

AI/ML Example
=    

Understanding extreme values can help with:

Outlier detection 
    
Risk modeling
    
Feature analysis

Empirical Rule — 68–95–99.7
=

For approximately normally distributed data:

68%

Approximately 68% of observations fall within:


    \mu \pm 1\sigma 
95%

Approximately 95% fall within:

    \mu \pm 2\sigma 
99.7%

Approximately 99.7% fall within:

μ±3σ

```python
import numpy as np

data = np.random.normal(100, 10, 10000)

mean = np.mean(data)
std = np.std(data)

print("Mean:", mean)
print("Standard Deviation:", std)
```

Notebook 5 – Probability
=

Introduction to Probability
=

Probability measures the likelihood that an event will happen.

Range:

 0< P(A)< 1 

Where:

0 = impossible
1 = certain

Events
=

An event is an outcome or group of outcomes we are interested in.

Example:

Rolling a die:
=
Sample Space = {1,2,3,4,5,6}

Event of getting an even number:
=

A = {2,4,6}

Therefore:-
P(A)=6/3=0.5

Sample Space
=

The sample space contains all possible outcomes.

Example:
=    

Coin:
=

S = {Head, Tail}

Dice:
=

S = {1,2,3,4,5,6}

Independent Events
=

Two events are independent if one event does not affect the other.

Formula:
=
    

 P(A\cap B)=P(A)P(B) 

Example
=

Toss a coin
Roll a die

The coin result does not affect the die result.


    P(A \cap B) = P(A)P(B)

Dependent Events
=

Events are dependent when the occurrence of one event affects the probability of another.

Example:
=

A bag contains:

3 Red balls

2 Blue balls

If one ball is removed without replacement, the probability of the next draw changes.

Conditional Probability
=

Conditional probability means the probability of an event given that another event has already occurred.

Formula:
=

P(A|B) = \frac{P(A \cap B)}{P(B)}

Example

Suppose:

P(A and B) = 0.3
 P(B) = 0.6

Bayes' Theorem
=

Bayes' theorem calculates the probability of a hypothesis based on available evidence.

P(A|B) = \frac{P(B|A)P(A)}{P(B)}

AI/ML Example
=

Suppose a spam filter receives an email containing suspicious words.

Bayes' theorem can help calculate:
=

Probability that email is spam
given the observed words.

Applications:
=

Spam detection

Medical diagnosis

Classification

Recommendation systems

Probability Rules
=
Rule 1 — Complement
=
 P(A')=1-P(A) 

Example:
=

If:

 P(A)=0.7 

Then:

    P(A')=0.3 
Rule 2 — Addition


For mutually exclusive events:

 P(A\cup B)=P(A)+P(B) 
 
Rule 3 — Multiplication

For independent events:

 P(A\cap B)=P(A)P(B) 

Random Variables
=

A random variable assigns numerical values to outcomes.

Discrete Random Variable

Takes countable values.

Example:
=    

Number of customers

Number of defective products

Number of calls

Continuous Random Variable

Can take any value within a range.

Example:
=

Height

Weight

Temperature

Time

Expected Value
=

Expected value is the probability-weighted average of possible outcomes.

Formula:
=

\mathbb{E}[X] = \sum_x xP(X=x)


Example
=

Suppose:
=
| Outcome | Probability |
| ------- | ----------: |
| ₹100    |         0.5 |
| ₹200    |         0.5 |

Then:

 E(X)=(100)(0.5)+(200)(0.5) 
       E(X)=150

```python
outcomes = [100, 200]
probabilities = [0.5, 0.5]

expected_value = sum(
    x * p for x, p in zip(outcomes, probabilities)
)

print(expected_value)
```

Notebook 6 – Probability Distributions
=

Bernoulli Distribution
=
Definition
=    

Bernoulli distribution represents a single experiment with only two possible outcomes:

Success = 1

Failure = 0

Examples:
=

Pass / Fail
    
Yes / No

Click / No Click

Fraud / Not Fraud

Formula

                    P(X=x)=p x(1−p)1−x


where:
=

x = 0 or 1

p = probability of success

Real-world Example
=

A customer either purchases a product or does not.

AI/ML Example
=

Binary classification:
=    

Spam = 1

Not Spam = 0

```python
import numpy as np
import matplotlib.pyplot as plt

p = 0.6

x = [0, 1]
probabilities = [1-p, p]

plt.bar(x, probabilities)
plt.xticks([0, 1], ["Failure", "Success"])
plt.title("Bernoulli Distribution")
plt.xlabel("Outcome")
plt.ylabel("Probability")
plt.show()
```

Binomial Distribution
=
Definition
=

Binomial distribution counts the number of successes in a fixed number of independent Bernoulli trials.

Conditions:
=    

Fixed number of trials

Two outcomes

Independent trials

Same probability of success

Real-world Example

Out of 10 customers, how many purchase a product?

AI/ML Example
=

Predicting the number of successful outcomes among a fixed number of independent trials.

```python
from scipy.stats import binom
import numpy as np
import matplotlib.pyplot as plt

n = 10
p = 0.5

x = np.arange(0, n + 1)
probabilities = binom.pmf(x, n, p)

plt.bar(x, probabilities)
plt.title("Binomial Distribution")
plt.xlabel("Number of Successes")
plt.ylabel("Probability")
plt.show()
```

Poisson Distribution
=
Definition
=    

Poisson distribution models the number of events occurring within a fixed interval of time or space

Where:
=

\(\lambda\) = average number of events

\(k\) = number of events

Real-world Examples
=
Number of customers arriving per hour

Number of calls per minute

Number of website requests per second

Number of accidents per month

AI/ML Example
=

It can be used to model count-based data such as:

Number of user interactions
Number of events
Number of failures

```python
from scipy.stats import poisson
import numpy as np
import matplotlib.pyplot as plt

lam = 4

x = np.arange(0, 15)
probabilities = poisson.pmf(x, lam)

plt.bar(x, probabilities)
plt.title("Poisson Distribution")
plt.xlabel("Number of Events")
plt.ylabel("Probability")
plt.show()
```

Uniform Distribution
=
Definition
=    

In a continuous uniform distribution, every value between \(a\) and \(b\) has the same density.

Real-world Example
=

A random number generated between 0 and 10.

AI/ML Example
=

Uniform distributions are commonly used when initializing certain model parameters or generating random samples.



```python
from scipy.stats import uniform
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 1000)
y = uniform.pdf(x, loc=0, scale=10)

plt.plot(x, y)
plt.title("Uniform Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

Normal Distribution
=
Definition
=
Normal distribution is a continuous, symmetric, bell-shaped distribution.

Parameters:
=    

Mean = μ

Standard deviation = σ

Real-world Example
=

Measurements such as heights or measurement errors can sometimes be approximately normally distributed.

AI/ML Example
=    

Normal distributions are important in:

Statistical modeling
    
Gaussian noise
    
Regression assumptions
    
Probabilistic models
    
Feature analysis

```python
from scipy.stats import norm
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(-4, 4, 1000)
y = norm.pdf(x, 0, 1)

plt.plot(x, y)
plt.title("Normal Distribution")
plt.xlabel("Value")
plt.ylabel("Density")
plt.show()
```

Exponential Distribution
=
Definition
=    

Exponential distribution models the waiting time until an event occurs when events happen continuously at a constant average rate.

Real-world Examples
=    
Time until next customer arrives
    
Time until machine failure
    
Waiting time for a phone call
    
Time between website requests

AI/ML Example
=
Used for modeling:

Time-to-event data

Failure analysis

Survival analysis

Waiting-time processes
    
    

```python
from scipy.stats import expon
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 10, 1000)
y = expon.pdf(x, scale=2)

plt.plot(x, y)
plt.title("Exponential Distribution")
plt.xlabel("Waiting Time")
plt.ylabel("Density")
plt.show()
```
