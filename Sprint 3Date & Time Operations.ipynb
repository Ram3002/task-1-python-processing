Notebook 11 – Date & Time Operations
=


Datetime
=    

Pandas provides pd.to_datetime() to convert values into datetime format.

```python
import pandas as pd

dates = pd.Series(["2026-01-10", "2026-02-15", "2026-03-20"])

dates = pd.to_datetime(dates)

print(dates)
```

Timestamp
=

A Timestamp represents a specific date and time.

```python
date = pd.Timestamp("2026-10-08 09:30:00")

print(date)
print(date.year)
print(date.month)
print(date.day)
```

Date Parsing
=

Date parsing means converting a string into a proper datetime value.

```python
df = pd.DataFrame({
    "date": ["08-10-2026", "15-10-2026", "20-10-2026"]
})

df["date"] = pd.to_datetime(df["date"], format="%d-%m-%Y")

print(df)
```

Timedelta
=

Timedelta represents a difference between two dates or times.

```python
start = pd.Timestamp("2026-10-01")
end = pd.Timestamp("2026-10-08")

difference = end - start

print(difference)
```

Adding days
=

```python
date = pd.Timestamp("2026-10-08")

print(date + pd.Timedelta(days=5))
```

Time Zones
=

Time zones allow us to work with dates and times from different locations.

```python
date = pd.Timestamp("2026-10-08 09:00:00", tz="Asia/Kolkata")

print(date)
```

Convert to another timezone:
=

```python
print(date.tz_convert("US/Eastern"))
```

Extracting Date Features
=

We can extract year, month, day, weekday, etc.

```python
df = pd.DataFrame({
    "date": pd.to_datetime([
        "2026-01-10",
        "2026-02-15",
        "2026-03-20"
    ])
})

df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["weekday"] = df["date"].dt.day_name()

print(df)
```

Notebook 12 – String Operations
=

lower()
=

Converts text to lowercase.

```python
df = pd.DataFrame({
    "name": ["RAM", "KUMAR", "PYTHON"]
})

df["name"] = df["name"].str.lower()

print(df)
```

upper()
=

Converts text to uppercase.

```python
df["name"] = df["name"].str.upper()

print(df)
```

strip()
=

Removes spaces from the beginning and end.

```python
df = pd.DataFrame({
    "name": ["  Ram  ", " Kumar ", " Python"]
})

df["name"] = df["name"].str.strip()

print(df)
```

split()
=

Splits a string into multiple parts.

```python
df = pd.DataFrame({
    "name": ["Ram Kumar", "John Smith", "Anu Raj"]
})

df[["first_name", "last_name"]] = df["name"].str.split(" ", expand=True)

print(df)
```

contains()
=
Checks whether a string contains a specific value.

```python
df = pd.DataFrame({
    "name": ["Ram", "Kumar", "Ravi", "John"]
})

result = df[df["name"].str.contains("Ra")]

print(result)
```

replace()
=

Replaces text with another value.

```python
df = pd.DataFrame({
    "city": ["Chennai", "Chennai", "Bangalore"]
})

df["city"] = df["city"].str.replace("Chennai", "Madurai")

print(df)
```

Regular Expressions (Regex)
=

Regex is used to find or validate patterns in text.

Example – Find emails

```python
df = pd.DataFrame({
    "email": [
        "ram@gmail.com",
        "kumar@yahoo.com",
        "invalid-email"
    ]
})

pattern = r"^[\w\.-]+@[\w\.-]+\.\w+$"

result = df[df["email"].str.contains(pattern, regex=True)]

print(result)
```

Notebook 13 – Missing Values
=

isnull()
=

Checks for missing values.

```python
df = pd.DataFrame({
    "Name": ["Ram", "Kumar", None],
    "Age": [25, None, 30]
})

print(df.isnull())
```

notnull()
=
Checks for values that are not missing.

```python
print(df.notnull())
```

dropna()
=

Removes rows containing missing values.

```python
df = pd.DataFrame({
    "Name": ["Ram", "Kumar", "Ravi"],
    "Age": [25, None, 30]
})

result = df.dropna()

print(result)
```

fillna()
=
Replaces missing values with a specified value.

```python
df = pd.DataFrame({
    "Name": ["Ram", "Kumar", "Ravi"],
    "Age": [25, None, 30]
})

df["Age"] = df["Age"].fillna(0)
```

interpolate()
=
Interpolation estimates missing values based on surrounding values.

```python
df = pd.DataFrame({
    "Sales": [100, 200, None, 400, 500]
})

df["Sales"] = df["Sales"].interpolate()

print(df)
```

Missing Value Handling Strategies
=

Different situations require different strategies.

| Strategy                | When to Use                            |
| ----------------------- | -------------------------------------- |
| `dropna()`              | When few rows have missing data        |
| `fillna(0)`             | When zero is meaningful                |
| Mean                    | Numerical data without strong outliers |
| Median                  | Numerical data with outliers           |
| Mode                    | Categorical data                       |
| Forward Fill (`ffill`)  | Time-series data                       |
| Backward Fill (`bfill`) | Time-series data                       |
| `interpolate()`         | Continuous/time-series values          |
