Notebook 2 – Measures of Central Tendency
=

Measures of central tendency describe the central or typical value of a dataset.

The main measures are:
=   

Mean

Median

Mode

Weighted Mean

Geometric Mean

Harmonic Mean


Mean
=
Definition
=
The mean is the average of all values in a dataset.


Numerical Example

Dataset:

10, 20, 30, 40, 50


The mean tells the company the average monthly sales.

AI/ML Example
=    

Suppose a dataset contains customer ages:

20, 25, 30, 35, 40

The mean age helps summarize the typical age of customers.

Numerical Example – Odd

10, 20, 20, 30, 40

Weighted Mean
=
Definition
=

A weighted mean gives different importance or weight to different observations.

   

Weighted mean provides the employee's overall performance score.

AI/ML Example
=

Weighted averages are used when some observations are more important than others.

For example, a model may assign higher weights to certain training examples.

Interpretation
=

Weighted mean is useful when all observations do not have equal importance.    


Geometric Mean
=
Definition
=    

The geometric mean is the nth root of the product of n values.

Geometric mean is useful for calculating average growth rates over multiple periods.

AI/ML Example
=

Geometric mean can be useful when combining ratios, growth rates, or multiplicative effects.

For example, evaluating multiplicative changes across multiple time periods.

Interpretation
=    

Geometric mean is particularly useful for growth rates and ratios.    

Harmonic Mean
=
Definition
=

The harmonic mean is useful for averaging rates and ratios, particularly when the quantities share a common numerator.
    
Business Example
=

Suppose a vehicle travels the same distance at:

The average speed for equal distances is calculated using the harmonic mean.

AI/ML Example
=                

Harmonic mean is important in precision and recall, where it produces the F1-score    

Interpretation
=    

Harmonic mean gives more importance to smaller values and is useful for rates.   

| Measure        | Main Use             | Sensitive to Outliers? |
| -------------- | -------------------- | ---------------------- |
| Mean           | Average              | Yes                    |

| Mode           | Most frequent value  | No                     |

Notebook 3 – Measures of Dispersion
=

Measures of dispersion tell us how spread out the data is.

Topics:
=

Range

Variance

Standard Deviation

Quartiles

Percentiles

Interquartile Range

Coefficient of Variation

Range
=
Definition
=

Range is the difference between the maximum and minimum values.

Formula
=    
                  Range=Maximum−Minimum

AI/ML Example
=

Range can provide a quick understanding of the spread of a feature.

Interpretation
=    

A larger range indicates a larger overall spread, although range depends only on two observations.

Variance
=
Definition
=

Variance measures the average squared distance of observations from the mean.

Business Example

A company can calculate variance in monthly sales to determine how consistently sales are performing.

AI/ML Example
=    

Variance is important in understanding feature variability and is also fundamental to concepts such as variance in statistical models.

Interpretation

Higher variance means observations are generally more spread out from the mean.    

Standard Deviation
=
Definition
=    

Standard deviation is the square root of variance

Business Example
=

Suppose two stores have the same average daily sales:

Store A → Average = ₹50,000
Store B → Average = ₹50,000

If Store A has a smaller standard deviation, its sales are more consistent.

AI/ML Example
=    

Standard deviation is commonly used when standardizing features.    

Interpretation
=    
Low SD → values are close to the mean.
High SD → values are widely spread.    

Quartiles
=
Definition
=

Quartiles divide ordered data into four parts.

There are three important quartiles:
=

Q1 → 25th percentile

Q2 → 50th percentile / Median

Q3 → 75th percentile

Numerical Example

10, 20, 30, 40, 50, 60, 70

Business Example
=

A company can divide customers into quartiles based on spending:

Q1 → lower-spending group

Q2 → middle

Q3 → higher-spending group

AI/ML Example
=

Quartiles can help understand the distribution of numerical features and identify potential outliers.

Interpretation
=   

Quartiles help understand where observations are positioned within a dataset.


Percentiles
=
Definition
=

A percentile indicates the value below which a given percentage of observations fall.

For example:
=    

90th percentile means approximately 90% of observations are at or below that value.

Numerical Example

 10, 20, 30, 40, 50

Business Example

An e-commerce company could examine the 90th percentile of delivery time.

If P90 is 5 days, approximately 90% of deliveries take 5 days or less under the chosen percentile convention.

AI/ML Example
=

Percentiles are commonly used for:

Outlier detection

Feature analysis

Data preprocessing

Understanding distributions

Interpretation
=

Percentiles tell us the relative position of a value in a dataset.


Interquartile Range – IQR
=
Definition
=    

IQR measures the spread of the middle 50% of data.

    IQR=Q3−Q1

Numerical Example

Suppose:

Q1 = 20
Q3 = 60

     IQR=60−20=40


Business Example
=

A company can use IQR to understand the typical variation in:

Customer spending

Delivery time

Order value

Employee salary

AI/ML Example
=

IQR is frequently used for outlier detection.

Common rule:
=
 Lower\ Bound = Q1 - 1.5(IQR) 
 Upper\ Bound = Q3 + 1.5(IQR) 

Values outside these bounds are often flagged as potential outliers.

Interpretation
=

IQR focuses on the middle 50% and is less affected by extreme values than the range.

Coefficient of Variation – CV
=
Definition
=

Coefficient of Variation measures standard deviation relative to the mean.

It is useful for comparing variability between datasets with different scales or means.

Business Example

Suppose two investments have:

Investment A

Mean return = 10%

SD = 2%

Investment B

Mean return = 20%

SD = 5%

his provides a relative measure of variability.

AI/ML Example
=

CV can help compare variability of features that have different means and scales.

Interpretation
=    

A higher CV means greater variability relative to the mean.


| Measure            | What it tells us             |
| ------------------ | ---------------------------- |
| Range              | Overall spread               |
| Variance           | Average squared deviation    |
| Standard Deviation | Typical spread around mean   |
| Quartiles          | Data divided into four parts |
| Percentiles        | Relative position            |
| IQR                | Spread of middle 50%         |
| CV                 | Relative variability         |
