Foreign Key
=

A Foreign Key connects one table to another table.

Suppose we have:

Departments

    departments.department_id
        ↑
        |
students.department_id




Why use it?

It maintains referential integrity and prevents invalid references.

```python
CREATE TABLE departments (
    department_id INT PRIMARY KEY,
    department_name VARCHAR(100)
)
```

```python
CREATE TABLE students (
    student_id INT PRIMARY KEY,
    name VARCHAR(100),
    department_id INT,
    FOREIGN KEY (department_id)
        REFERENCES departments(department_id)
);
```

Window Functions
=

Window functions perform calculations across related rows without collapsing the rows into one row per group.

This is the key difference from GROUP BY

ROW_NUMBER()
=

ROW_NUMBER() assigns a unique sequential number to each row within a window.

Example:

| name  | salary | row_num |
| ----- | -----: | ------: |
| Ravi  |  60000 |       1 |
| Anu   |  55000 |       2 |
| Arun  |  50000 |       3 |
| Priya |  45000 |       4 |
| Kumar |  40000 |       5 |
    

RANK()
=

RANK() gives the same rank to tied values and leaves gaps after ties

DENSE_RANK()
=
DENSE_RANK() also gives the same rank to ties, but does not leave gaps.

For:

| Function       | Salaries            | Result  |
| -------------- | ------------------- | ------- |
| `ROW_NUMBER()` | 60000, 60000, 50000 | 1, 2, 3 |
| `RANK()`       | 60000, 60000, 50000 | 1, 1, 3 |
| `DENSE_RANK()` | 60000, 60000, 50000 | 1, 1, 2 |
